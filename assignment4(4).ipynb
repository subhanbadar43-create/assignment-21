```python
def calculate_bill(amount, discount_type="normal"):
    if amount >= 10000:
        discount_percent = 20
    elif amount >= 5000:
        discount_percent = 10
    elif amount >= 2000:
        discount_percent = 5
    else:
        discount_percent = 0
    if discount_type == "festival":
        discount_percent += 5
    discount_amount = (amount * discount_percent) / 100
    final_price = amount - discount_amount
    print(f"Original Price : {amount}")
    print(f"Discount ({discount_percent}%) : {discount_amount}")
    print(f"Final Price : {final_price}")
    print("-" * 30)
    return final_price
calculate_bill(12000)
calculate_bill(6000)
calculate_bill(2500)
calculate_bill(1000)
calculate_bill(12000, discount_type="festival")
```
